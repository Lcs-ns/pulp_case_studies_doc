```python
import pulp as pl

Oils = ["VEG1", "VEG2", "OIL1", "OIL2", "OIL3"]

costs = {
    "VEG1": 110,
    "VEG2": 120,
    "OIL1": 130,
    "OIL2": 110,
    "OIL3": 115
}

hardness = {
    "VEG1": 8.8,
    "VEG2": 6.1,
    "OIL1": 2.0,
    "OIL2": 4.2,
    "OIL3": 5.0
}
```

```python
prob = pl.LpProblem("Vegetable and Non-Vegetable Oils Problem", pl.LpMaximize)
```

```python
oil_vars = pl.LpVariable.dicts("oil", Oils, 0)
```

```python
prob += pl.lpSum(costs[i] * oil_vars[i] for i in Oils) == 150, "PercentageSum"
```
